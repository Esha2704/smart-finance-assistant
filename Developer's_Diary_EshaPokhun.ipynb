```python
import pandas as pd

# Read the Excel file
df = pd.read_excel("sample_transactions for business programming.xlsx")

# Show first 5 rows
print(df.head())
```

```python
# Remove dollar signs from Amount column
df["Amount"] = (
    df["Amount"]
    .str.replace("$", "", regex=False)
)

# Convert Amount into numbers
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")

# Check results
print(df.head())
```

```python
# Calculate total spending by category
category_spending = (
    df.groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

# Display results
print("TOTAL SPENDING BY CATEGORY\n")
print(category_spending)
```

```python
# Identify highest spending category
top_category = category_spending.idxmax()

print("SMART FINANCIAL INSIGHTS\n")

# Business recommendations based on spending patterns
if top_category == "Groceries":
    print("Groceries are your highest expense category.")
    print("Consider meal planning and bulk purchasing to reduce costs.\n")

if category_spending.get("Coffee", 0) > 30:
    print("Coffee spending is relatively high.")
    print("Home brewing could help increase monthly savings.\n")

if category_spending.get("Entertainment", 0) > 150:
    print("Entertainment expenses are significant.")
    print("Consider setting a monthly entertainment budget.\n")

if category_spending.get("Refund", 0) < 0:
    print("Refund transactions were detected.")
    print("Your assistant successfully handled negative transaction values.\n")

# Overall summary
total_spending = category_spending.sum()

print(f"Total Recorded Spending: ${total_spending:.2f}")
```

```python
import matplotlib.pyplot as plt

# Create spending chart
category_spending.plot(kind="bar")

# Add chart title and labels
plt.title("Total Spending by Category")
plt.xlabel("Category")
plt.ylabel("Amount Spent ($)")

# Display chart
plt.show()
```
