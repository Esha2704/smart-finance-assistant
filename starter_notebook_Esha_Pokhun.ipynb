### Foundation Data Processing Functions

````python
# 🤖 AI Collaboration: Data Loading and Cleaning Function
# Ask AI to help you create a robust data loading function

def load_and_clean_transaction_data(data_source):
    """
    Load and clean transaction data for the Smart Finance Assistant.

    This function handles:
    - CSV files or existing DataFrames
    - Dollar signs in Amount values
    - Missing values
    - Invalid dates or amounts
    - Required column validation
    """

    try:
        # Load data depending on input type
        if isinstance(data_source, pd.DataFrame):
            transaction_data = data_source.copy()
        else:
            transaction_data = pd.read_csv(data_source)

        # Required columns for the finance assistant
        required_columns = ['Date', 'Amount', 'Category', 'Description']

        # Check if all required columns exist
        for column in required_columns:
            if column not in transaction_data.columns:
                raise ValueError(f"Missing required column: {column}")

        # Remove rows where all values are missing
        transaction_data = transaction_data.dropna(how='all')

        # Fill missing category and description values
        transaction_data['Category'] = transaction_data['Category'].fillna('Uncategorised')
        transaction_data['Description'] = transaction_data['Description'].fillna('No description provided')

        # Clean Amount column by removing dollar signs and commas
        transaction_data['Amount'] = (
            transaction_data['Amount']
            .astype(str)
            .str.replace('$', '', regex=False)
            .str.replace(',', '', regex=False)
            .str.strip()
        )

        # Convert Amount to numeric
        transaction_data['Amount'] = pd.to_numeric(transaction_data['Amount'], errors='coerce')

        # Remove rows with invalid amount values
        transaction_data = transaction_data.dropna(subset=['Amount'])

        # Convert Date column to datetime
        transaction_data['Date'] = pd.to_datetime(transaction_data['Date'], errors='coerce', dayfirst=True)

        # Remove rows with invalid dates
        transaction_data = transaction_data.dropna(subset=['Date'])

        # Create a transaction type column
        transaction_data['Transaction Type'] = transaction_data['Amount'].apply(
            lambda amount: 'Refund' if amount < 0 else 'Spending'
        )

        print("✅ Transaction data loaded and cleaned successfully!")
        print(f"Total valid transactions: {len(transaction_data)}")

        return transaction_data

    except Exception as error:
        print(f"❌ Error loading transaction data: {error}")
        return None
    # Your AI-assisted implementation goes here
    # Replace this placeholder with AI-generated code
    pass

# Test your function
# clean_data = load_and_clean_transaction_data(df_sample)
# print(clean_data)
````

````python
# 🤖 AI Collaboration: Spending Analysis Function
# Ask AI to help you create comprehensive spending analysis

def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights.
    """

    try:
        if df is None or df.empty:
            raise ValueError("No transaction data available for analysis.")

        # Separate spending and refunds
        spending_data = df[df['Amount'] > 0]
        refund_data = df[df['Amount'] < 0]

        # Main totals
        total_spending = spending_data['Amount'].sum()
        total_refunds = refund_data['Amount'].sum()
        net_spending = df['Amount'].sum()
        average_transaction = spending_data['Amount'].mean()

        # Spending by category
        category_summary = (
            spending_data
            .groupby('Category')['Amount']
            .agg(['sum', 'mean', 'count'])
            .reset_index()
        )

        category_summary.columns = [
            'Category',
            'Total Spent',
            'Average Transaction',
            'Number of Transactions'
        ]

        # Calculate percentage of total spending
        category_summary['Percentage of Spending'] = (
            category_summary['Total Spent'] / total_spending * 100
        )

        # Sort highest spending categories first
        category_summary = category_summary.sort_values(
            by='Total Spent',
            ascending=False
        )

        highest_category = category_summary.iloc[0]['Category']

        analysis_results = {
            'total_spending': total_spending,
            'total_refunds': total_refunds,
            'net_spending': net_spending,
            'average_transaction': average_transaction,
            'highest_spending_category': highest_category,
            'category_summary': category_summary
        }

        print("✅ Spending analysis completed successfully!")
        print(f"Total Spending: ${total_spending:.2f}")
        print(f"Total Refunds: ${total_refunds:.2f}")
        print(f"Net Spending: ${net_spending:.2f}")
        print(f"Average Transaction: ${average_transaction:.2f}")
        print(f"Highest Spending Category: {highest_category}")

        return analysis_results

    except Exception as error:
        print(f"❌ Error analysing spending patterns: {error}")
        return None
````

````python
analysis = analyze_spending_patterns(clean_data)

print("\n📊 Category Spending Summary:")
print(analysis['category_summary'])
````

````python
# 🤖 AI Collaboration: Business Insights Generator
# Ask AI to help create professional financial recommendations

def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis.
    """

    try:
        if analysis_data is None:
            raise ValueError("No analysis data available.")

        category_summary = analysis_data['category_summary']
        highest_category = analysis_data['highest_spending_category']
        total_spending = analysis_data['total_spending']
        average_transaction = analysis_data['average_transaction']
        net_spending = analysis_data['net_spending']

        recommendations = []

        recommendations.append("💰 SMART FINANCE ASSISTANT REPORT")
        recommendations.append("=" * 40)
        recommendations.append(f"Total spending: ${total_spending:.2f}")
        recommendations.append(f"Net spending after refunds: ${net_spending:.2f}")
        recommendations.append(f"Average transaction: ${average_transaction:.2f}")
        recommendations.append(f"Highest spending category: {highest_category}")
        recommendations.append("")

        recommendations.append("📊 Key Insights:")

        top_row = category_summary.iloc[0]
        recommendations.append(
            f"- Your highest spending category is {top_row['Category']} "
            f"with ${top_row['Total Spent']:.2f}, representing "
            f"{top_row['Percentage of Spending']:.1f}% of total spending."
        )

        # Category-specific recommendations
        for _, row in category_summary.iterrows():
            category = row['Category']
            total = row['Total Spent']
            percentage = row['Percentage of Spending']

            if percentage >= 30:
                recommendations.append(
                    f"- {category} takes up a large share of your spending "
                    f"({percentage:.1f}%). Consider setting a monthly limit."
                )

            if category == "Coffee" and row['Number of Transactions'] >= 4:
                recommendations.append(
                    "- Coffee purchases are frequent. A weekly coffee budget "
                    "could help reduce small daily expenses."
                )

            if category == "Entertainment" and total > 150:
                recommendations.append(
                    "- Entertainment spending is relatively high. Consider choosing "
                    "lower-cost activities occasionally."
                )

            if category == "Dining" and total > 100:
                recommendations.append(
                    "- Dining expenses are noticeable. Preparing some meals at home "
                    "could help save money."
                )

        recommendations.append("")
        recommendations.append("✅ Recommended Action:")
        recommendations.append(
            "Focus first on the highest spending category because reducing even a "
            "small percentage there can create meaningful savings."
        )

        return "\n".join(recommendations)

    except Exception as error:
        return f"❌ Error generating recommendations: {error}"
````

````python
# 🤖 AI Collaboration: Custom Tool Development
# Ask AI to help you create useful financial calculators

from hands_on_ai import agent

def create_savings_calculator_tool(current_savings, monthly_contribution, target_amount):
    """
    Calculate how many months are needed to reach a savings goal.
    """

    try:
        current_savings = float(current_savings)
        monthly_contribution = float(monthly_contribution)
        target_amount = float(target_amount)

        if current_savings < 0 or monthly_contribution <= 0 or target_amount <= 0:
            raise ValueError("Savings values must be positive, and monthly contribution must be greater than zero.")

        remaining_amount = target_amount - current_savings

        if remaining_amount <= 0:
            return "✅ You have already reached your savings goal!"

        months_needed = remaining_amount / monthly_contribution

        return (
            "💰 Savings Goal Calculator\n"
            "===========================\n"
            f"Current savings: ${current_savings:.2f}\n"
            f"Target amount: ${target_amount:.2f}\n"
            f"Monthly contribution: ${monthly_contribution:.2f}\n"
            f"Remaining amount: ${remaining_amount:.2f}\n"
            f"Estimated time to reach goal: {months_needed:.1f} months"
        )

    except Exception as error:
        return f"❌ Error calculating savings goal: {error}"
````

````python
result = create_savings_calculator_tool(
    current_savings=500,
    monthly_contribution=150,
    target_amount=2000
)

print(result)
````
